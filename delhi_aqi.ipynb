ENCODING DATA 

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
stan = StandardScaler()
la_season = LabelEncoder()
la_city = LabelEncoder()
la_aqi_category = LabelEncoder()
df['city_encoded'] = la_city.fit_transform(df['city'])
df['season_encoded'] = la_season.fit_transform(df['season'])
df['aqi_category_encoded'] = la_aqi_category.fit_transform(df['aqi_category'])
df['datetime'] = df['datetime'].astype('int64')
df['datetime'] = stan.fit_transform(df[['datetime']])
```

MODEL FITTING

```python
from sklearn.linear_model import LogisticRegression
X = df[['datetime','is_weekend','season_encoded','city_encoded','no2','so2','co','o3','temperature','humidity','wind_speed','visibility','aqi']]
y = df['aqi_category_encoded']
model = LogisticRegression()
model.fit(X,y)
```

MODEL PREDICTION

```python
datetime = input('enter datetime (YYYY-MM-DD TT:TT:TT): ')
datetime = (pd.to_datetime(datetime)).value
datetime = stan.transform([[datetime]])
datetime = datetime[0][0]
season = int(input("enter the season in no:" \
"for winter: 3" \
"for post_monsoon: 1" \
"for monsoon: 0" \
"for summer: 2" \
"enter: "))
city = int(input("enter the city in no:" \
"for faridabad: 1" \
"for delhi: 0" \
"for ghaziabad: 2" \
"for gurugram: 3" \
"for noida: 4" \
"enter : "))
is_weekend = int(input('is there a weekend that day:'))
no2 = float(input('enter no2: '))
so2 = float(input('enter so2:'))
co = float(input('enter co:'))
o3 = float(input('enter o3:'))
temperature = float(input('enter the temperature: '))
humidity = float(input('enter humidity'))
wind_speed = float(input('enter wind speed: '))
visibility = float(input('enter visibility: '))
aqi = float(input('enter aqi:'))
answer = model.predict([[datetime,is_weekend,season,city,no2,so2,co,o3,temperature,humidity,wind_speed,visibility,aqi]])
answer 

```

```python
answered = answer[0]
answered
```

```python
answered_category = df[df['aqi_category_encoded']==answered]['aqi_category'].drop_duplicates()
```

```python
answered_category= answered_category[0]
```
